- Demonstrate deposits, withdrawals, and interest application.

````python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance = self.balance + amount

    def withdraw(self, amount):
        self.balance = self.balance - amount

class SavingsAccount(BankAccount):
    def __init__(self, owner, balance, interest_rate):
        super().__init__(owner, balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        self.balance = self.balance + (self.balance * self.interest_rate)

chisom_account = SavingsAccount('Chisom Alor', 5000, 0.05)
chisom_account.deposit(1000)
print("After deposit:", chisom_account.balance)

chisom_account.withdraw(500)
print("After withdrawal:", chisom_account.balance)

chisom_account.add_interest()
print("After interest:", chisom_account.balance)

````

- Take a list of numbers `[10, 20, 30, 40, 50]`.  
- Convert it into an iterator using `iter()` and retrieve each element using `next()`.  
- Handle the `StopIteration` exception gracefully when there are no more items.

````python
numbers = [10, 20, 30, 40, 50]
num_iterator = iter(numbers)

# Retrieve each element manually using next() in a loop
while True:
    try:
        current_val = next(num_iterator)
        print(f"Retrived number: {current_val}")
    except StopIteration:
        print("Reached the end of the list! StopIteration caught gracefully")
        break
````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
class CountDown:
    def __init__(self, n):
        self.current = n

    def __iter__(self):
        return self

    def __next__(self):
        if self.current >= 1:
            val = self.current
            self.current -= 1
        else:
            raise StopIteration

# Demonstrate in a for loop
print("Countdown to week 2 Submission:")
for num in CountDown(5):
    print(f"Step {num}...")
````

- Create a custom iterator `FileLineReader` that takes a filename and reads one line at a time.  
- Implement `__iter__()` and `__next__()` so it can be used in a `for` loop.  
- Test it by reading from a sample `.txt` file and printing each line.

````python
class FileLineReader:
    def __init__(self, filename):
        self.file = open(filename, 'r')

    def __iter__(self):
        return self

    def __next__(self):
        line = self.file.readline()
        if line == "":
            self.file.close()
            raise StopIteration
        return line.strip()

# First, create a sample file to test with
with open("sample.txt", "w") as f:
    f.write("Chisom Alor\n")
    f.write("Plant Science and Biotechnology\n")
    f.write("DataraFlow Week 2\n")

# Now test the FileLineReader
reader = FileLineReader("sample.txt")
for line in reader:
    print(line)
````

- Create an abstract class `Shape` with an abstract method `area()`.  
- Create subclasses `Rectangle` and `Circle` that implement the `area()` method.  
- Create a list of different shapes and loop through them, printing their area without checking their type.


````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * (self.radius ** 2)

# Create list of different shape object
shapes = [Rectangle(10, 5), Circle(4), Rectangle(3, 7)]

# Loop through and print their area without checking object types
for shape in shapes:
    print(f"Calculated Shape Area: {shape.area():.2f}")
        
````

Create two classes: `Flyer` (method `move()` returns `"Flying"`) and `Swimmer` (method `move()` returns `"Swimming"`).  
Create a class `Duck` inheriting from both.  
- Use `super()` to decide which `move()` executes.  
- Explain the **Method Resolution Order (MRO)**.  

Method Resolution Order (MRO) —
When a class inherits from multiple parents, Python needs a clear RULE for which parent to check FIRST if there’s a naming conflict. That rule is called the Method Resolution Order — and it simply follows the ORDER i listed the parents in.
Since i wrote class Duck(Flyer, Swimmer): — Flyer is listed FIRST, Python checks Flyer BEFORE Swimmer. So super().move() finds Flyer’s version first, and that’s what runs — hence the output “Flying.”


````python
class Flyer:
    def move(self):
        return "Flying"

class Swimmer:
    def move(self):
        return "Swimming"

class Duck(Flyer, Swimmer):
    def move(self):
        return super().move()

d1 = Duck()
print(d1.move())
````

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
class Fibonacci:
    def __init__(self, n):
        self.n = n
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration
        value = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return value

for number in Fibonacci(10):
    print(number)
````

````python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def get_description(self):
        return f"{self.title} by {self.author} ({self.year})"

class EBook(Book):
    def __init__(self, title, author, year, file_size, format):
        super().__init__(title, author, year)
        self.file_size = file_size
        self.format = format

    def get_description(self):
        return f"{self.title} by {self.author} ({self.year}) - {self.format}, {self.file_size}MB"

class PrintedBook(Book):
    def __init__(self, title, author, year, weight, dimensions):
        super().__init__(title, author, year)
        self.weight = weight
        self.dimensions = dimensions

    def get_description(self):
        return f"{self.title} by {self.author} ({self.year}) - {self.weight}g, {self.dimensions}"
````

Part 2:Member, StudentMember, TeacherMember

````python
class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self._borrowed_books = []   # protected

    def borrow_book(self, book):
        self._borrowed_books.append(book)

    def __iter__(self):
        return iter(self._borrowed_books)

class StudentMember(Member):
    def __init__(self, name, member_id, grade_level):
        super().__init__(name, member_id)
        self.grade_level = grade_level
        self.borrow_limit = 3

    def borrow_book(self, book):
        if len(self._borrowed_books) < self.borrow_limit:
            self._borrowed_books.append(book)
            print(f"{self.name} borrowed {book.title}")
        else:
            print(f"{self.name} has reached their borrow limit of {self.borrow_limit}")

class TeacherMember(Member):
    def __init__(self, name, member_id, department):
        super().__init__(name, member_id)
        self.department = department
        self.borrow_limit = 10

    def borrow_book(self, book):
        if len(self._borrowed_books) < self.borrow_limit:
            self._borrowed_books.append(book)
            print(f"{self.name} borrowed {book.title}")
        else:
            print(f"{self.name} has reached their borrow limit of {self.borrow_limit}")
````

Part 3:Library class, with private name + iteration + tracking

````python
class Library:
    total_books_borrowed = 0   # class-level variable, shared across the whole system

    def __init__(self, name):
        self.__library_name = name
        self.books = []

    def get_library_name(self):
        return self.__library_name

    def set_library_name(self, new_name):
        self.__library_name = new_name

    def add_book(self, book):
        self.books.append(book)

    def search_by_title(self, title):
        return [book for book in self.books if title.lower() in book.title.lower()]

    def search_by_author(self, author):
        return [book for book in self.books if author.lower() in book.author.lower()]

    def __iter__(self):
        return iter(self.books)

    @classmethod
    def record_borrow(cls):
        cls.total_books_borrowed += 1
````

Part 4: Putting it all togerther

````python
library = Library("Chisom's DataraFlow Library")

book1 = EBook("Things Fall Apart", "Chinua Achebe", 1958, 15, "PDF")
book2 = PrintedBook("Plant Physiology", "Elvis & Zoe", 2015, 1200, "20x25cm")
book3 = EBook("Data Science Handbook", "Emma Okeke", 2016, 20, "EPUB")

library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

print("All books in library:")
for book in library:
    print(book.get_description())

student = StudentMember("Ada", "S001", "300 Level")
teacher = TeacherMember("Dr. Nwosu", "T001", "Plant Science")

student.borrow_book(book1)
student.borrow_book(book2)
Library.record_borrow()
Library.record_borrow()

teacher.borrow_book(book3)
Library.record_borrow()

print("\nAda's borrowed books:")
for book in student:
    print(book.get_description())

print("\nSearching for 'Python':", [b.title for b in library.search_by_title("Python")])
print("Total books borrowed across the library:", Library.total_books_borrowed)
print("Library name (via getter):", library.get_library_name())
````
