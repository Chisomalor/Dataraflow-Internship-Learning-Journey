Assignment 2

````python
import random

score = 0

for i in range(5):
    num1 = random.randint(1, 20)
    num2 = random.randint(1, 20)

    operation = random.choice(["+", "-", "*"])
    if operation == "+":
        correct_answer = num1 + num2
    elif operation == "-":
        answer = num1 - num2
    else:
        correct_answer = num1 * num2

    user_answer = int(input(f"What is {num1} {operation} {num2}? "))

    if user_answer == answer:
        print("Correct!")
        score += 1
    else:
        print(f"Incorrect. The correct answer is {correct_answer}.")

print(f"\nYou scored {score} out of 5.")
````

Assignment 4

````python
import datetime

birthday = input("Enter your birthday (YYYY-MM-DD): ")
birthday_date = datetime.datetime.strptime(birthday, "%Y-%m-%d")

today = datetime.datetime.now()
next_birthday = birthday_date.replace(year=today.year)

if next_birthday < today:
    next_birthday = next_birthday.replace(year=today.year + 1)

days_until_birthday = (next_birthday - today).days
print(f"There are {days_until_birthday} days until your next birthday.")
````

Assessment 5: What is pip?

pip is Python's package installer.

It is used to download and install external libraries/packages that are not built into Python by default, from the Python Package Index (PyPI). This lets developers add extra functionality to their projects, like `requests` for making internet requests, or `numpy` for numerical computing, without having to write that functionality themselves.

Example command:
```
pip install requests
```

What happens opening a JSON file with invalid content?

Python raises a json.JSONDecodeError, which means the file's text doesn't follow proper JSON structure (like a missing comma or bracket), so it can't be parsed into Python object.

Example:

````python
import json

invalid_json = '{"name": "Chisom", "age": }'

try:
    data = json.loads(invalid_json)
    print(data)

except json.JSONDecodeError:
    print("Invalid JSON content.")
````

Assessment 14: The command to install a specific version of a package using pip

You use two equal signs (`==`) after the package name, followed by the exact version number you want. This tells pip to install that specific version, rather than authomatically grabbing the newest one avaliable.

Example code:
```
pip install requests==2.31.0
```
This is useful when a project needs a specific version for compactibility, for example, if a newer version of a package have changed something that breaks existing code, pinning an exact version avoids that problem.
